```python
from ultralytics import YOLO

# Load the pre-trained YOLOv10n model
model = YOLO("yolov10n.pt")

# Perform object detection on an image
results = model("/kaggle/input/dibetic-retinopathy-2-v2i-yolov9/valid/images/dr-1002_png.rf.46a01679358a6fd371ff60683813b027.jpg")

# Display the results
results[0].show()
```

```python
from ultralytics import YOLO

# Load the model (YOLOv10n as a starting point)
model = YOLO("yolov10n.pt")

# Train the model on your custom dataset
model.train(
    data="/kaggle/input/dibetic-retinopathy-2-v2i-yolov9/data.yaml",  # Replace with your actual data.yaml path
    epochs=50,
    imgsz=640,
    batch=16,
    workers=4
)
```
